```python
import os
import cv2
import numpy as np
import scipy
from PIL import Image
```

```python
def extract_instances(mat_file, output_dir):
    filename = os.path.splitext(os.path.basename(mat_file))[0]

    # Load the matrix file
    mat_data = scipy.io.loadmat(mat_file)

    inst_map = mat_data["inst_map"]

    # Convert instance segmentation to binary gland mask
    gland_mask = (inst_map > 0).astype(np.uint8) * 255

    # Save as black/white PNG
    Image.fromarray(gland_mask).save(f"{output_dir}/{filename}.jpg")

    return
```

```python
mat_file = "../Image_10646.mat"
output_dir = "../datasets/healthy2im/maskB"
mat_data = scipy.io.loadmat(mat_file)

extract_instances(mat_file, output_dir)
```

```python
from pathlib import Path
MAT_DIR = "../datasets/healthy2im/gland_segmented_trainB/gland_mat"
OUTPUT_DIR = "../datasets/healthy2im/maskB"

# .iterdir() gets everything; .is_file() filters out folders
for file in Path(MAT_DIR).iterdir():
    if file.is_file():
        extract_instances(file.absolute(), output_dir)
```

```python
from pathlib import Path
import shutil

dir1 = Path("../datasets/healthy2im/trainB")
dir2 = Path("../datasets/healthy2im/maskB")
output_dir = Path("../datasets/healthy2im/matched_trainB")

output_dir.mkdir(parents=True, exist_ok=True)

# Get all file stems in dir2
dir2_stems = {file.stem for file in dir2.iterdir() if file.is_file()}

# Check files in dir1
for file in dir1.iterdir():
    if file.is_file() and file.stem in dir2_stems:
        shutil.copy2(file, output_dir / file.name)
        print(f"Copied: {file.name}")

print("Done!")
```
