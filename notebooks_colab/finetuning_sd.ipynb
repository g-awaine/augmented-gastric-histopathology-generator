```python
from pathlib import Path
import json
DIR = "../datasets/healthy2im/maskB"


with open("dataset.jsonl", "w", encoding="utf-8") as f:
    for file in Path(DIR).iterdir():
        if file.is_file():
            entry = {"file_name": file.name, "text": "intestinal metaplasia"}
            # json.dumps converts dict to a string, then we append a newline character
            f.write(json.dumps(entry) + "\n")
```
